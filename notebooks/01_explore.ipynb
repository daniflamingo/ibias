# First look at the Guardian corpus

Run `ibias fetch`, `ibias load` and `ibias count` first. Each cell below is a question you can ask the data; change them freely.

```python
import duckdb
from ibias import analyse, config

con = duckdb.connect(str(config.DB_PATH), read_only=True)
```

## How many articles, and from which pillars?

DuckDB lets you write plain SQL against the `articles` table. `.df()` turns the result into a pandas DataFrame.

```python
con.sql("""
    SELECT pillar, count(*) AS articles, round(avg(word_count)) AS avg_words
    FROM articles GROUP BY pillar ORDER BY articles DESC
""").df()
```

## Most common words and phrases

```python
con.close()  # analyse opens its own connection
counts = analyse.load_counts("pillar")
analyse.top_terms(counts)
```

## Which phrases are typical of Opinion pieces?

This log-odds comparison is the core of the Gentzkow & Shapiro method. In stage 1b the two groups will be Labour and Conservative MPs instead of Opinion and everything else.

```python
analyse.distinctive_terms(counts, "Opinion", n=2)
```
